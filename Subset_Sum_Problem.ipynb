```python
A = [7, 40, 45, 49, 42, 34, 16, 18, 48, 17, 19, 47, 5, 43, 29, 20, 30, 44, 26, 26, 50, 8, 17, 15, 21, 23, 17, 24, 41, 41, 34, 10, 11, 36, 43, 43, 18, 11, 1, 42, 5, 8, 39, 22, 2, 6, 18, 14, 25, 26]
target = 900
n = len(A)
H = 0
x = [Binary('x'+str(i)) for i in range(n)]
```

```python
for i in range(n):
  H += A[i]*x[i]
H -= target
H = H*H
```

```python
model = H.compile()
bqm = model.to_bqm()
sampler = neal.SimulatedAnnealingSampler()
samplest = sampler.sample(bqm,num_reads=2000)
```

```python
if samplest.first.energy != 0:
  print("no feasible solution")
  sys.exit()
```

```python
one_of_the_best_solutions = samplest.first.sample
sorted_one_of_the_best_solutions = sorted(one_of_the_best_solutions.keys(), key=lambda k: int(k[1:]))
print("one of the best solutions: (", end="")
for k in sorted_one_of_the_best_solutions:
  if k != sorted_one_of_the_best_solutions[-1]:
    print(f"{k}: {int(one_of_the_best_solutions[k])},", end="")
  else:
    print(f"{k}: {int(one_of_the_best_solutions[k])}", end="")
print(")")
print("Hamiltonian: ", samplest.first.energy)
```
