```python
G = nx.read_edgelist(io.StringIO(keller4_data), nodetype=int)
edges_list = list(G.edges())
n = 171
H = 0
x = [Binary('x'+str(i)) for i in range(1,n+1)]
```

$
Minimize \ \ \ \ A\sum_{i=1}^{n} \sum_{j=1}^{n} q_{ij}(1-x_i)(1-x_j) \ \ \ + \ \ \ B\sum_{i=1}^{n}x_i
\\ Subject \ To: \ x \ \in \ \{0,1\}^{n} \  
\\ 和以下的寫法等價   
\\ Minimize \ \ \ \ A\sum_{i=1}^{n} \sum_{j=1}^{n} q_{ij}x_ix_j \ \ \ - \ \ \ B\sum_{i=1}^{n}x_i \ + \ Bn
\\ Subject \ To: \ x \ \in \ \{0,1\}^{n} \
$

$ G = (V,E) \ ，\  \ V=\{1,2,...,n\} \ 是節點集合 \ ， \ E是邊集合
\\ \begin{cases}
q_{ij} = 1, \ \ (i,j) \ \in \ E
\\ q_{ij} = 0, \ \ otherwise
\end{cases}
\\ 看起來設定 \ A=1,B=1 \ 是可以的，但保險起見設定 \ A=5,B=1 \ $

```python
for edge in edges_list:
  H += x[edge[0]-1]*x[edge[1]-1]
H *= 5
for i in range(n):
  H -= x[i]*x[i]
```

```python
model = H.compile()
bqm = model.to_bqm()
sampler = neal.SimulatedAnnealingSampler()
samplest = sampler.sample(
    bqm,
    num_reads=5000,
    sweeps=10000,
)
```

```python
one_of_the_best_solutions = samplest.first.sample
sorted_one_of_the_best_solutions = sorted(one_of_the_best_solutions.keys(), key=lambda k: int(k[1:]))
print("one of the best solutions: (", end="")
for k in sorted_one_of_the_best_solutions:
  if k != sorted_one_of_the_best_solutions[-1]:
    print(f"{k}: {1-int(one_of_the_best_solutions[k])},", end="")
  else:
    print(f"{k}: {1-int(one_of_the_best_solutions[k])}", end="")
print(")")
print("Hamiltonian: ", samplest.first.energy+n)
```
